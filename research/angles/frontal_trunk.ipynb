```python
import os
import math
from decimal import Decimal
import pandas as pd
from google.colab import files
import io
```

```python
uploaded = files.upload()

all_data = pd.DataFrame()
```

```python
def mid(x1, y1, x2, y2):
    midx = ((x1 + x2)/2)
    midy = ((y1 + y2)/2)
    return midx, midy
```

```python
def side1(x1, x2):
    side = math.sqrt(((x1[0] - x2[0])**2) + ((x1[1] - x2[1])**2))
    return side


def side2(x1, x2, y2):
    side = math.sqrt(((x1[0] - x2)**2) + ((x1[1] - y2)**2))
    return side


def cosB(a, b, c):
    cos = (a**2 + c**2 - b**2) / (2 * a * c)
    return cos
```

```python
for file_name in uploaded.keys():
    if file_name.endswith(".csv"):

        df = pd.read_csv(io.BytesIO(uploaded[file_name]))

        all_data = pd.concat([all_data, df], ignore_index=True)
```

```python
kpts_columns = all_data['kpts'].str.split(',', expand=True)


for i in range(kpts_columns.shape[1]):
    kpts_columns.rename(columns={i: f'kpt_{i+1}'}, inplace=True)


all_data = all_data.join(kpts_columns)
```

```python
x5 = all_data['kpt_13']
y5 = all_data['kpt_14']
x6 = all_data['kpt_16']
y6 = all_data['kpt_17']
x11 = all_data['kpt_31']
y11 = all_data['kpt_32']
x12 = all_data['kpt_34']
y12 = all_data['kpt_35']
```

```python
top_mid = []
bot_mid = []
a_side = []
b_side = []
c_side = []
an = []


for i in range(len(x5)):
    top_ = mid(Decimal(x5[i]), Decimal(y5[i]), Decimal(x6[i]), Decimal(y6[i]))
    bot_ = mid(Decimal(x11[i]), Decimal(y11[i]), Decimal(x12[i]), Decimal(y12[i]))
    a_ = side1(top_, bot_)
    b_ = side2(top_, Decimal(x12[i]), Decimal(y12[i]))
    c_ = side2(bot_, Decimal(x12[i]), Decimal(y12[i]))
    an_ = cosB(a_, b_, c_)

    top_mid.append(top_)
    bot_mid.append(bot_)
    a_side.append(a_)
    b_side.append(b_)
    c_side.append(c_)
    an.append(math.degrees(math.acos(an_)))
```

```python
all_data['an'] = an
```

```python
print(an)
```

```python
average_angle = sum(an) / len(an)
print(f"平均脊椎角度: {average_angle}")
```

```python
all_data.to_csv('脊椎角度.csv', index=False)


files.download('脊椎角度.csv')
```
