```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
import io
import math
```

```python
uploaded = files.upload()


file_name = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[file_name]))
```

```python
def angle_change(x1, y1, x2, y2, x3, y3):

    a_x = x2 - x1
    a_y = y2 - y1

    b_x = x3 - x2
    b_y = y3 - y2

    dot_product = a_x * b_x + a_y * b_y

    magnitude_a = math.sqrt(a_x**2 + a_y**2)
    magnitude_b = math.sqrt(b_x**2 + b_y**2)
    if magnitude_a * magnitude_b == 0:
        return 0

    cos_angle = dot_product / (magnitude_a * magnitude_b)

    cos_angle = max(-1, min(1, cos_angle))

    return math.acos(cos_angle) * (180 / math.pi)
```

```python
print(f"Columns in the file: {df.columns.tolist()}")
```

```python
x12 = df['kpt_34'].astype(float)
y12 = df['kpt_35'].astype(float)
x13 = df['kpt_37'].astype(float)
y13 = df['kpt_38'].astype(float)
x14 = df['kpt_40'].astype(float)
y14 = df['kpt_41'].astype(float)
x15 = df['kpt_44'].astype(float)
y15 = df['kpt_45'].astype(float)
x18 = (x12 + x13) / 2
y18 = (y12 + y13) / 2
```

```python
angle_changes = []
for i in range(len(x18)):
    angle_change_value = angle_change(x12[i], y12[i], x18[i], y18[i], x13[i], y13[i])
    angle_changes.append(angle_change_value)
```

```python
df['angle_change'] = angle_changes


df['delta_t'] = df['time'].diff()
df['delta_theta'] = df['angle_change'].diff()


df['angular_velocity'] = df['delta_theta'] / df['delta_t']
```
