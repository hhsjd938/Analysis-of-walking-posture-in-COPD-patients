```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
import io
```

```python
print("請上傳 Q1 對應的文件:")
uploaded_q1 = files.upload()
print("請上傳 Q2 對應的文件:")
uploaded_q2 = files.upload()
print("請上傳 Q3 對應的文件:")
uploaded_q3 = files.upload()
```

```python
def read_csv(uploaded):
    for file_name in uploaded.keys():
        df = pd.read_csv(io.BytesIO(uploaded[file_name]))
        return df
```

```python
df_q1 = read_csv(uploaded_q1)
df_q2 = read_csv(uploaded_q2)
df_q3 = read_csv(uploaded_q3)
```

```python
q1_angles = df_q1['Q1_angle_change']
q2_angles = df_q2['Q2_angle_change']
q3_angles = df_q3['Q3_angle_change']
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot([q1_angles, q2_angles, q3_angles], labels=['Q1', 'Q2', 'Q3'], patch_artist=True, boxprops=dict(facecolor='skyblue'))
plt.title('Q1, Q2, Q3 Angle Change Box Plot')
plt.xlabel('Quartile')
plt.ylabel('Angle Change')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(q1_angles, bins=30, alpha=0.5, label='Q1', color='blue')
plt.hist(q2_angles, bins=30, alpha=0.5, label='Q2', color='green')
plt.hist(q3_angles, bins=30, alpha=0.5, label='Q3', color='red')
plt.title('Q1, Q2, Q3 Angle Change Histogram')
plt.xlabel('Angle Change')
plt.ylabel('Frequency')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(q1_angles, label='Q1', color='blue', marker='o', markersize=2)
plt.plot(q2_angles, label='Q2', color='green', marker='o', markersize=2)
plt.plot(q3_angles, label='Q3', color='red', marker='o', markersize=2)
plt.title('Q1, Q2, Q3 Angle Change Line Plot')
plt.xlabel('Data Points')
plt.ylabel('Angle Change')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
