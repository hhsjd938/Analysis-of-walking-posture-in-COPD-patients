```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
import io
```

```python
print("請上傳 Q1 對應的文件:")
uploaded_q1 = files.upload()

print("請上傳 Q2 對應的文件:")
uploaded_q2 = files.upload()

print("請上傳 Q3 對應的文件:")
uploaded_q3 = files.upload()
```

```python
q1_file = list(uploaded_q1.keys())[0]
q2_file = list(uploaded_q2.keys())[0]
q3_file = list(uploaded_q3.keys())[0]

all_data_q1 = pd.read_csv(io.BytesIO(uploaded_q1[q1_file]))
all_data_q2 = pd.read_csv(io.BytesIO(uploaded_q2[q2_file]))
all_data_q3 = pd.read_csv(io.BytesIO(uploaded_q3[q3_file]))
```

```python
q1_angles = all_data_q1['Q1_angle_change'].tolist()
q2_angles = all_data_q2['Q2_angle_change'].tolist()
q3_angles = all_data_q3['Q3_angle_change'].tolist()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(q1_angles, label='Q1', marker='o', linestyle='-', color='blue')
plt.plot(q2_angles, label='Q2', marker='o', linestyle='-', color='green')

plt.title('Q1, Q2, Q3 Angle Change Line Plot')
plt.xlabel('Data Points')
plt.ylabel('Angle Change')
plt.legend()
plt.tight_layout()
plt.show()
```
