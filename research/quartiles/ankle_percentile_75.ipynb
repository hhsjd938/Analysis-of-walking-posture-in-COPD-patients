```python
import os
import math
from decimal import Decimal
import pandas as pd
from google.colab import files
import io
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
```

```python
uploaded = files.upload()


all_data = pd.DataFrame()
```

```python
for file_name in uploaded.keys():
    if file_name.endswith(".csv"):

        df = pd.read_csv(io.BytesIO(uploaded[file_name]))

        all_data = pd.concat([all_data, df], ignore_index=True)
```

```python
kpts_columns = all_data['kpts'].str.split(',', expand=True)
```

```python
for i in range(kpts_columns.shape[1]):
    kpts_columns.rename(columns={i: f'kpt_{i+1}'}, inplace=True)


all_data = all_data.join(kpts_columns, rsuffix='_kpts')
```

```python
def step_x(x1, x2):
    step_ans = abs(x2 - x1)
    return step_ans
```

```python
x16 = all_data['kpt_46'].astype(float)
x17 = all_data['kpt_49'].astype(float)
```

```python
step_lengths = [step_x(x16[i], x17[i]) for i in range(len(x16))]
```

```python
sorted_step_lengths = sorted(step_lengths)
```

```python
lower_cutoff_index = int(len(sorted_step_lengths) * 0.25)
```

```python
top_75_percent_steps = sorted_step_lengths[lower_cutoff_index:]
```

```python
total_step_length_75 = sum(top_75_percent_steps)
average_step_length_75 = sum(top_75_percent_steps) / len(top_75_percent_steps) if top_75_percent_steps else 0
```

```python
print(f"中間75%數據的總步距: {total_step_length_75}")
print(f"中間75%數據的平均步距: {average_step_length_75}")
```

```python
all_data.to_csv('75%的步距.csv', index=False)
files.download('75%的步距.csv')
```
